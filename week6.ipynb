```python
import pandas as pd
import numpy as np

pd.set_option('display.max_rows', None)
pd.set_option('display.max_columns', None)
pd.set_option('display.float_format', '{:.3f}'.format)
pd.set_option('display.width', 1000)

```

```python
#i needed to make it parquet because changing datetime in week 4-5 means nothing if it's not saved

sold = pd.read_parquet('data/sold_prepared.parquet')
listing = pd.read_parquet('data/listing_prepared.parquet')
```

```python
sold.dtypes
```

```python
#columns will be working with for key metrics:
# ClosePrice, OriginalListPrice, LivingArea, DaysOnMarkey, CloseDate, PurchaseContractDate, ListingContractDate



#print(f"Sold Columns: {sold.columns}") 
#print(f"Listing Columns: {listing.columns}")

columns_to_check = ['ClosePrice', 'OriginalListPrice', 'LivingArea', 'DaysOnMarket', 'CloseDate', 'PurchaseContractDate', 'ListingContractDate']

for col in columns_to_check:
    if col in sold:
        print(f"{col} in Sold")
    if col not in sold:
        print(f"{col} not in Sold")
    if col in listing: 
        print(f"{col} in Listing")
    if col not in listing:
        print(f"{col} not in Listing")


#all columns in dfs


for df in [sold,listing]: 
    df['price_ratio'] = df['ClosePrice']/ df['OriginalListPrice'] #same as capturing full price reduction history
    df['price_per_square_feet'] = df['ClosePrice']/ df['LivingArea']
    df['close_date_year'] = df['CloseDate'].dt.year
    df['close_date_month'] = df['CloseDate'].dt.month
    df['close_date_year_month'] = df['CloseDate'].dt.strftime('%Y-%m')
    df['listing_to_contract_days'] = (df['PurchaseContractDate'] - df['ListingContractDate']).dt.days
    df['contract_to_close_days'] = (df['CloseDate'] - df['PurchaseContractDate']).dt.days

#in listing, metrics may be null because it may not be sold yet

```

```python
#SEGEMENT ANALYSIS

#Sold - helps with understanding what has happened already

#grouping by property type
pricing_ratio_by_prop = sold.groupby(['PropertyType', 'PropertySubType'])[['price_ratio', 'price_per_square_feet']].agg(['count','mean','median','min','max'])
#print(pricing_ratio_by_prop)
market_period_by_prop = sold.groupby(['PropertyType', 'PropertySubType'])[['DaysOnMarket','listing_to_contract_days', 'contract_to_close_days']].agg(['count','mean','median','min','max'])
#print(market_period_by_prop)


#grouping by area region
pricing_ratio_by_region = sold.groupby(['CountyOrParish', 'MLSAreaMajor'])[['price_ratio', 'price_per_square_feet']].agg(['count','mean','median','min','max'])
#print(pricing_ratio_by_region)
market_period_by_region = sold.groupby(['CountyOrParish', 'MLSAreaMajor'])[['DaysOnMarket','listing_to_contract_days', 'contract_to_close_days']].agg(['count','mean','median','min','max'])
#print(market_period_by_region)

#grouping by competition
pricing_ratio_list_office = sold.groupby('ListOfficeName')[['price_ratio', 'price_per_square_feet']].agg(['count','mean','median','min','max'])
market_period_list_office = sold.groupby('ListOfficeName')[['DaysOnMarket','listing_to_contract_days', 'contract_to_close_days']].agg(['count','mean','median','min','max'])
#print(pricing_ratio_list_office)
#print(market_period_list_office)
pricing_ratio_buyer_office = sold.groupby('BuyerOfficeName')[['price_ratio', 'price_per_square_feet']].agg(['count','mean','median','min','max'])
market_period_buyer_office = sold.groupby('BuyerOfficeName')[['DaysOnMarket','listing_to_contract_days', 'contract_to_close_days']].agg(['count','mean','median','min','max'])
#price(pricing_ratio_buyer_office)
#price(market_period_buyer_office)



#Listing - wouldn't help much since the deals haven't closed in the dataset
#pricing_ratio_by_prop_listing = listing.groupby(['PropertyType', 'PropertySubType'])[['price_ratio', 'price_per_square_feet']].agg(['count','mean','median','min','max'])
#print(pricing_ratio_by_prop_listing)
#market_period_by_prop_listing = listing.groupby(['PropertyType', 'PropertySubType'])[['DaysOnMarket','listing_to_contract_days', 'contract_to_close_days']].agg(['count','mean','median','min','max'])
#print(market_period_by_prop_listing)


```
